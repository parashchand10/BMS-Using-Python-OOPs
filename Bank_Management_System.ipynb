# **Bank Management System - OOPs**



A simple Bank Management System built using Python **Object-Oriented Programming (OOP)** concepts.

This project demonstrates core OOP principles such as classes, inheritance, encapsulation, and data abstraction through real-world banking operations.

The system is designed to provide a clear and structured understanding of how OOP is applied in practical applications.

```python
class Bank: # Parent Class
  def __init__(self, name, age, city):  # init method
    self.name = name
    self.age = age
    self.city = city    # properties

  def customer_details(self):
    print(f"Name: {self.name}")
    print(f"Age: {self.age}")
    print(f"City: {self.city}")
```

```python
accounts = set() # Empty Set
```

```python
class Account_Holder(Bank):                   # Child Class
  def __init__(self, name, age, city, account_number, balance):
      super().__init__(name, age, city)
      self.balance = balance                  # Account Balance
      self.account_number = account_number    # Customer Account Number

  def acc_balance(self):
      print(f"Account balance: {self.balance}")

  def __acc_operations(self):
      choice = input("Enter your choice (w, d, t): ")

      if choice == "w":        # Withdraw
          withdraw_amt = int(input("Enter the withdraw amount: "))

          if withdraw_amt <= self.balance:
              self.balance = self.balance - withdraw_amt
              print(f"Your current balance is: {self.balance}")
          else:
              print("!Not enough balance")

      elif choice == "d":      # Deposite
          deposite_amt = int(input("Enter the deposite amount: "))
          self.balance = self.balance + deposite_amt
          print(f"Your current balance is: {self.balance}")

      elif choice == "t":      # Transfer
          receiver_acc = int(input("Enter receiver account number: "))
          transfer_amt = int(input("Enter transfer amount: "))

          for trans_amt in accounts:
            if trans_amt.account_number == receiver_acc:
                receiver = trans_amt
                break

          self.balance = self.balance - transfer_amt            # deduct amount from sender
          receiver.balance = receiver.balance + transfer_amt    # add amount from receiver

          print("Transfer successful")
          print(f"Your balance: {self.balance}")

      else:
          print("Invalid choice")

  def get_acc_operations(self):     # Access private method
      self.__acc_operations()

  def get_detail(self):
      super().customer_details()
```

```python
acc1 = Account_Holder("Rahul", 25, "Delhi", 101, 5000)
acc2 = Account_Holder("Amit", 28, "Mumbai", 102, 3000)

accounts.add(acc1)
accounts.add(acc2)
```

```python
acc1.get_detail() # Customer Details
```

```python
acc1.get_acc_operations()
```

```python
acc1.balance  # Check account balance
```
